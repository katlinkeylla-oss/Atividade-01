## 🥈 Questão 2 — A cotação do dólar (1,0)

**Área:** Câmbio · **Conceito:** indexação, fatiamento e `IndexError`


O financeiro registrou o fechamento do dólar (R$) em sete dias úteis:

```python
dolar = (5.42, 5.47, 5.39, 5.51, 5.58, 5.55, 5.61)
```

**Tarefa:** em um único código, produza:

1. o primeiro, o último, o penúltimo e o quarto valor (índice `3`);
2. os **três primeiros** (`dolar[0:3]`) e os **três últimos** (`dolar[-3:]`);
3. os valores de **dois em dois** (`dolar[::2]`) e a série **invertida** (`dolar[::-1]`);
4. o **tipo** de uma fatia (`type(...)`) e o tamanho da série;
5. a **variação percentual** do primeiro para o último dia;
6. uma consulta **segura** ao índice `10`: use `if indice < len(dolar)` e, na versão com `try`/`except IndexError`, imprima uma mensagem em vez de deixar o erro derrubar o notebook.

**Saída esperada:** primeiro **5.42**; último **5.61**; penúltimo **5.55**; quarto **5.51**; primeiros três `(5.42, 5.47, 5.39)`; últimos três `(5.58, 5.55, 5.61)`; de dois em dois `(5.42, 5.39, 5.58, 5.61)`; invertida `(5.61, 5.55, 5.58, 5.51, 5.39, 5.47, 5.42)`; tipo da fatia **tuple**; tamanho **7**; variação **3,51%**; o índice 10 **não existe**.

**Dica:** o fatiamento de uma tupla devolve outra **tupla**, nunca uma lista. Em `dolar[0:3]` o índice 3 fica de fora.

````python
dolar = (5.42, 5.47, 5.39, 5.51, 5.58, 5.55, 5.61)

# TODO: elementos pedidos
primeiro = dolar[0]
ultimo = dolar[-1]
penultimo = dolar[-2]
quarto = dolar[3]

# TODO: fatias
primeiros_tres = dolar[0:3]
ultimos_tres = dolar[-3:]
de_dois_em_dois = dolar[::2]
invertida = dolar[::-1]

# TODO: tipo e tamanho da fatia

variacao =  (ultimo / primeiro - 1) * 100

print(f"Primeiro: {primeiro} | Último: {ultimo} | Penúltimo: {penultimo} | Quarto: {quarto}")
print(f"Primeiros três: {primeiros_tres}")
print(f"Últimos três: {ultimos_tres}")
print(f"De dois em dois: {de_dois_em_dois}")
print(f"Invertida: {invertida}")
print(f"Tipo da fatia: {type(primeiros_tres).__name__} | Tamanho da série: {len(dolar)}")
print(f"Variação no período: {variacao:.2f}%")

# TODO: consulta segura ao índice 10 (versão 1: com if e len)
indice = 10
if indice < len(dolar):
    print(f"dolar[{indice}] = {dolar[indice]}")
else:
    print(f"Índice {indice} não existe (a série tem {len(dolar)} valores).")

# TODO: consulta segura ao índice 10 (versão 2: com try/except IndexError)
try:
    print(f"dolar[{indice}] = {dolar[indice]}")
except IndexError:
    print(f"Índice {indice} não existe (a série tem {len(dolar)} valores).")
````

**Sua resposta:**

**Área:** Auditoria fiscal · **Conceito:** lote de tuplas, desempacotamento, `while` e lista de ocorrências

```python
    ("NF-5002", "98.765.432/0001-10", 64200.0, "PIX"),
    ("NF-5003", "45.678.901/0001-22", 31900.0, "PIX"),
    ("NF-5004", "11.222.333/0001-44", 87500.0, "PIX"),
    ("NF-5005", "55.666.777/0001-88",  9800.0, "Boleto"),
)
```

**Regra:** notas de valor **igual ou superior a R$ 50.000,00** precisam de revisão da diretoria.

**Tarefa:** com um `while` e desempacotamento `nota, cnpj, valor, forma = notas_fiscais[i]`:

1. some o **volume total** auditado;
2. quando `valor >= 50000.0`, imprima o alerta `🚨 REVISAR: NF-5002 | R$ 64,200.00` e guarde a tupla `(nota, valor)` na lista `revisar`;
3. some o total das notas pagas em **PIX** e conte quantas foram;
4. calcule a **participação do PIX** e a **participação das notas a revisar** no volume total;
5. prove a integridade do lote: tente `notas_fiscais[0][2] = 0.0` dentro de `try`/`except TypeError`.

**Saída esperada:** volume total **R$ 211.900,00**; alertas **NF-5002 (64.200,00)** e **NF-5004 (87.500,00)**; total a revisar **R$ 151.700,00** (**71,59%**); PIX **3** notas somando **R$ 183.600,00** (**86,64%**); a tentativa de alterar o lote gera **TypeError**.

**Dica:** `revisar` é uma **lista** (cresce com `append`) cujos itens são **tuplas** (registros que não mudam). Listas e tuplas se complementam.

````python

notas_fiscais = (
    ("NF-5001", "12.345.678/0001-90", 18500.0, "Boleto"),
    ("NF-5002", "98.765.432/0001-10", 64200.0, "PIX"),
    ("NF-5003", "45.678.901/0001-22", 31900.0, "PIX"),
    ("NF-5004", "11.222.333/0001-44", 87500.0, "PIX"),
    ("NF-5005", "55.666.777/0001-88", 9800.0, "Boleto"),
)

limite = 50000.0

volume_total = 0.0
total_revisar = 0.0
total_pix = 0.0
qtd_pix = 0
revisar = []

i = 0
while i < len(notas_fiscais):

    # Desempacotamento da nota
    nota, cnpj, valor, forma = notas_fiscais[i]

    # Acumula o volume total
    volume_total += valor

    # Verifica se precisa de revisão
    if valor >= limite:
        total_revisar += valor
        print(f"🚨 REVISAR: {nota} | R$ {valor:,.2f}")
        revisar.append((nota, valor))

    # Acumula os dados do PIX
    if forma == "PIX":
        total_pix += valor
        qtd_pix += 1

    i += 1

# Calcula as porcentagens
part_pix = (total_pix / volume_total) * 100
part_revisar = (total_revisar / volume_total) * 100

# Mostra os resultados
print(f"\nVolume total: R$ {volume_total:,.2f}")
print(f"A revisar: {revisar} -> R$ {total_revisar:,.2f} ({part_revisar:.2f}% do volume)")
print(f"PIX: {qtd_pix} notas -> R$ {total_pix:,.2f} ({part_pix:.2f}% do volume)")
````

## 🏆 Questão 10 — O painel de pedidos (1,0)

**Área:** Pedidos · **Conceito:** capstone com `namedtuple`


Para deixar o código legível, o time de vendas definiu os pedidos com `namedtuple`:

```python
from collections import namedtuple

Pedido = namedtuple("Pedido", ["id", "cliente", "produto", "qtd", "preco"])

pedidos = (
    Pedido(1001, "Lojas Maia",  "Notebook", 3, 3200.0),
    Pedido(1002, "Atacado PB",  "Monitor", 10,  950.0),
    Pedido(1003, "Rede Brasil", "Teclado", 25,  120.0),
    Pedido(1004, "Mercado Sul", "Notebook", 2, 3200.0),
    Pedido(1005, "Lojas Maia",  "Monitor",  4,  950.0),
)
```

**Tarefa:** com um `while` percorrendo `pedidos` e **acessando os campos pelo nome** (`pedido.qtd`, `pedido.preco`):

1. calcule o **faturamento de cada pedido** (`qtd * preco`) e o **faturamento total**;
2. calcule o **ticket médio** (faturamento total ÷ número de pedidos);
3. some as **unidades de Notebook** vendidas;
4. conte os pedidos da **Lojas Maia** e some quanto ela comprou;
5. encontre o **maior pedido** (guarde o próprio objeto `Pedido`) e imprima seu `id` e `cliente`;
6. guarde `(id, faturamento)` de cada pedido em uma lista `resumo` e converta-a para tupla;
7. tente `pedidos[0].preco = 1.0` dentro de `try`/`except AttributeError`.

**Saída esperada:** faturamentos **9600,00; 9500,00; 3000,00; 6400,00; 3800,00**; total **R$ 32.300,00**; ticket médio **R$ 6.460,00**; **5** notebooks vendidos; Lojas Maia com **2** pedidos somando **R$ 13.400,00**; maior pedido **1001 — Lojas Maia (R$ 9.600,00)**; a alteração gera **AttributeError**.

**Dica:** a `namedtuple` continua sendo uma tupla (imutável, desempacotável, com índices), mas com **nomes** que tornam o código autoexplicativo — `pedido.qtd` é muito mais claro que `pedido[3]`.

````python

from collections import namedtuple

Pedido = namedtuple("Pedido", ["id", "cliente", "produto", "qtd", "preco"])

pedidos = (
    Pedido(1001, "Lojas Maia", "Notebook", 3, 3200.0),
    Pedido(1002, "Atacado PB", "Monitor", 10, 950.0),
    Pedido(1003, "Rede Brasil", "Teclado", 25, 120.0),
    Pedido(1004, "Mercado Sul", "Notebook", 2, 3200.0),
    Pedido(1005, "Lojas Maia", "Monitor", 4, 950.0),
)

faturamento_total = 0.0
notebooks = 0
pedidos_maia = 0
total_maia = 0.0
maior_pedido = None
maior_valor = 0.0
resumo = []

i = 0
while i < len(pedidos):
    pedido = pedidos[i]

    # Faturamento do pedido
    faturamento = pedido.qtd * pedido.preco
    faturamento_total += faturamento
    resumo.append((pedido.id, faturamento))

    # Notebooks vendidos
    if pedido.produto == "Notebook":
        notebooks += pedido.qtd

    # Pedidos da Lojas Maia
    if pedido.cliente == "Lojas Maia":
        pedidos_maia += 1
        total_maia += faturamento

    # Maior pedido
    if maior_pedido is None or faturamento > maior_valor:
        maior_pedido = pedido
        maior_valor = faturamento

    print(f"Pedido {pedido.id} | {pedido.cliente:<12} | R$ {faturamento:>9,.2f}")
    i += 1

ticket_medio = faturamento_total / len(pedidos)
resumo = tuple(resumo)

print(f"\nFaturamento total: R$ {faturamento_total:,.2f} | Ticket médio: R$ {ticket_medio:,.2f}")
print(f"Notebooks vendidos: {notebooks}")
print(f"Lojas Maia: {pedidos_maia} pedidos | R$ {total_maia:,.2f}")
print(f"Maior pedido: {maior_pedido.id} — {maior_pedido.cliente} (R$ {maior_valor:,.2f})")
print(f"Resumo: {resumo}")

# Tentar alterar um campo
try:
    pedidos[0].preco = 1.0
except AttributeError:
    print("namedtuple também é imutável: não é possível alterar o campo.")

    print("namedtuple também é imutável: não é possível alterar o campo.")
````
