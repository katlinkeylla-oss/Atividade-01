## 🥉 Questão 1 — A abertura do boletim (1,0)

**Área:** Análise de mercado · **Conceito:** criação, indexação, fatiamento, `len` e cópia

A série de fechamento da ação **VARE3** nos últimos seis dias é:

```python
precos = [28.40, 29.10, 27.80, 30.20, 31.00, 30.50]
```

**Tarefa:** em um único código, produza:

1. o primeiro, o último, o terceiro e o penúltimo preço (use índices positivos e negativos);
2. os três primeiros e os três últimos preços, com fatiamento;
3. o tamanho da série com `len()`;
4. a correção do terceiro preço de `27.80` para `28.70` (altere a lista);
5. uma **cópia independente** `precos_copia = precos[:]`; altere `precos_copia[0] = 99.99` e mostre que `precos` não mudou.

**Saída esperada:** primeiro **28.4**; último **30.5**; terceiro **27.8**; penúltimo **31.0**; primeiros três `[28.4, 29.1, 27.8]`; últimos três `[30.2, 31.0, 30.5]`; `len = 6`; após a correção, `precos[2] == 28.7`; a cópia começa com **99.99** e o original continua com **28.4**.

**Dica:** `precos[:3]` inclui os índices 0, 1 e 2. O índice 3 fica de fora.

````python
precos = [28.40, 29.10, 27.80, 30.20, 31.00, 30.50]

# TODO: acesse os elementos pedidos
primeiro = precos[0]
ultimo = precos[-1]
terceiro = precos[2]
penultimo = precos[-2]

# TODO: fatie os três primeiros e os três últimos
primeiros_tres = precos[:3]
ultimos_tres = precos[-3:]

# TODO: use len()
quantidade = len(precos)

# TODO: corrija o terceiro preço
precos[2] = 28.7

# TODO: crie a cópia independente e altere o primeiro elemento dela
precos_copia = precos[:]
precos_copia[0] = 99.99

# TODO: imprima os resultados
print(f"Quantidade: {quantidade}")

print(f"Primeiro: {primeiro}")
print(f"Último: {ultimo}")
print(f"Terceiro: {terceiro}")
print(f"Penúltimo: {penultimo}")
print(f"Primeiros três: {primeiros_tres}")
print(f"Últimos três: {ultimos_tres}")
print(f"Quantidade: {quantidade}")
print(f"Série corrigida: {precos}")
print(f"Cópia alterada: {precos_copia}")
````

### ✍️ Explique (Q1)

Qual é a diferença entre `precos[-1]` e `precos[len(precos) - 1]`? E por que `precos_copia = precos[:]` é diferente de `precos_copia = precos`?

**Sua resposta:**

As duas formas conseguem pegar o último preço da lista, mas precos[-1] é mais simples porque já pega o último valor diretamente, sem precisar contar os elementos. Já na cópia, precos[:] cria uma lista separada da original. Assim, se eu mudar uma, a outra continua igual. Mas, usando precos_copia = precos, as duas usam a mesma lista, então qualquer alteração aparece nas duas.

A série de vendas da **Casa do Norte** chegou assim:

```python
vendas = [120, 135, 142, 150, 142]
```

O analista identificou:

- falta o mês de **março**, que vendeu **128** — ele deve entrar entre fevereiro (135) e abril (142);
- o último **142** é uma duplicata de abril e deve sair;
- **maio** foi digitado como **150**, mas o correto é **155**;
- falta **junho**, que vendeu **138**.

**Tarefa:** corrija a série **na ordem** e, ao final, imprima a lista, `len(vendas)`, `vendas.count(142)` e `vendas.index(142)`.

**Saída esperada:** `[120, 135, 128, 142, 155, 138]`, tamanho **6**, contagem de 142 = **1**, índice de 142 = **3**.

> 🧠 **Pergunta:** por que `vendas.remove(142)` não resolveria o problema da duplicata?

````python
vendas = [120, 135, 142, 150, 142]

# TODO: insira março (128) no índice correto
vendas.insert(2, 128)

# TODO: remova a duplicata do final guardando o valor em `duplicado`
duplicado = vendas.pop()

# TODO: corrija maio para 155
vendas[3] = 155

# TODO: acrescente junho (138)
vendas.append(138)

print(f"Série final: {vendas}")
print(f"Tamanho: {len(vendas)}")
print(f"Ocorrências de 142: {vendas.count(142)}")
print(f"Índice de 142: {vendas.index(142) if 142 in vendas else 'Não encontrado'}")
print(f"Valor removido como duplicata: {duplicado}")
````

## 🥇 Questão 3 — O ranking de produtos (1,0)

**Área:** Vendas · **Conceito:** `sort`, `index`, fatiamento e `while`

A **Casa do Norte** vende cinco linhas de produto:

```python
produtos = ["Essência", "Aurora", "Brisa", "Vértice", "Norte"]
receitas = [1500.0, 3200.0, 900.0, 2500.0, 1800.0]
```

**Tarefa:**

1. crie `ranking = receitas[:]` e ordene **decrescente** com `sort(reverse=True)`;
2. com um `while`, imprima o **top 3** no formato `1º Aurora — R$ 3200.00`, usando `receitas.index(ranking[i])` para recuperar o produto;
3. calcule a receita total com `while` e a **participação do líder** no total;
4. crie `ranking_asc = receitas[:]`, ordene crescente e imprima;
5. explique por que `receitas.sort()` diretamente **quebraria** a correspondência entre as duas listas.

**Saída esperada:** top 3 **Aurora (3200)**, **Vértice (2500)**, **Norte (1800)**; total **9.900,00**; participação do líder **32,32%**; ordem crescente `[900.0, 1500.0, 1800.0, 2500.0, 3200.0]`.

**Dica:** `ranking` guarda só os valores. Para descobrir o produto de um valor, use `receitas.index(valor)` — funciona porque não há receitas repetidas nesta base.

````python
produtos = ["Essência", "Aurora", "Brisa", "Vértice", "Norte"]
receitas = [1500.0, 3200.0, 900.0, 2500.0, 1800.0]

# TODO: crie e ordene o ranking decrescente
ranking = receitas[:]
ranking.sort(reverse=True)


# TODO: imprima o top 3 com while
i = 0
while i < 3:
    posicao = receitas.index(ranking[i])
    valor = ranking[i]
    produto = produtos[posicao]
    print(f"{i + 1}º {produto} — R$ {valor:.2f}")
    i += 1

# TODO: calcule a receita total com while e a participação do líder
total = 0.0
i = 0
while i < len(receitas):
    total += receitas[i]
    i += 1
participacao_lider = ranking[0] / total * 100

print(f"Receita total: R$ {total:.2f}")
print(f"Participação do líder: {participacao_lider:.2f}%")

# TODO: crie e ordene o ranking crescente
ranking_asc = receitas[:]
ranking_asc.sort()

print(f"Ordem crescente: {ranking_asc}")
````

### ✍️ Explique (Q3)

O que aconteceria com a **correspondência** entre `produtos` e `receitas` se você escrevesse `receitas.sort(reverse=True)` em vez de `ranking = receitas[:]` e ordenasse `ranking`? Por que isso é um problema para o boletim?

**Sua resposta:**
Se eu ordenasse a lista original, os valores mudariam de lugar, mas os nomes dos produtos continuariam na mesma ordem. Isso poderia misturar os valores e os produtos no boletim. Por isso, é melhor criar uma cópia para fazer a organização.

## 🥈 Questão 4 — O risco da VARE3 (1,0)

**Área:** Mercado de capitais · **Conceito:** listas + `while` + acumuladores

A série de preços da **VARE3** em sete dias:

```python
precos = [30.00, 31.20, 29.80, 32.50, 33.10, 31.90, 30.40]
```

**Tarefa:** usando `while` (não use `sum`, `max`, `min`), calcule:

- preço médio;
- maior e menor preço;
- amplitude (`maior - menor`);
- quantos dias o preço ficou **acima da média**;
- quantos dias tiveram **retorno negativo** (preço do dia menor que o do dia anterior).

Antes de calcular, valide a série: se `len(precos) == 0`, `raise ValueError`.

5. imprima a lista `reposicao` com um `while` (não use `for`).

3. construa `desvios` com `append` (`peso_atual - peso_alvo`);
4. com um `while`, imprima uma recomendação por ativo:
   - `REDUZIR` se `desvio > 0.02`;
   - `AUMENTAR` se `desvio < -0.02`;
   - `MANTER` caso contrário;
5. conte quantos ativos estão acima, abaixo e dentro da tolerância de **2 p.p.**

**Saída esperada:** valor total **R$ 14.760,00**; pesos atuais aproximadamente `[0.2608, 0.2100, 0.3848, 0.1443]`; desvios `[-0.0392, 0.0100, 0.0348, -0.0057]`; recomendações: PETR4 **AUMENTAR**, VALE3 **MANTER**, ITUB4 **REDUZIR**, BBDC4 **MANTER**; **1 acima**, **1 abaixo**, **2 dentro** da tolerância.

````python
ativos = ["PETR4", "VALE3", "ITUB4", "BBDC4"]
quantidades = [100, 50, 200, 150]
precos = [38.50, 62.00, 28.40, 14.20]
pesos_alvo = [0.30, 0.20, 0.35, 0.15]

# TODO: calcule o valor total com while
total = 0.0
i = 0
while i < len(quantidades):
    total += quantidades[i] * precos[i]
    i += 1

# TODO: construa pesos_atuais e desvios com append
pesos_atuais = []
desvios = []
i = 0
while i < len(quantidades):
    peso = quantidades[i] * precos[i] / total
    pesos_atuais.append(peso)
    desvios.append(peso - pesos_alvo[i])
    i += 1

# TODO: imprima as recomendações e conte as categorias
i = 0
acima = 0
abaixo = 0
dentro = 0

while i < len(ativos):
    if desvios[i] > 0.02:
        print(f"{ativos[i]}: AUMENTAR")
        acima += 1

    elif desvios[i] < -0.02:
        print(f"{ativos[i]}: REDUZIR")
        abaixo += 1

    else:
        print(f"{ativos[i]}: MANTER")
        dentro += 1

    i += 1

print(f"Valor total: R$ {total:.2f}")
print(f"Pesos atuais: {pesos_atuais}")
print(f"Desvios: {desvios}")
print(f"Acima: {acima} | Abaixo: {abaixo} | Dentro: {dentro}")
````
